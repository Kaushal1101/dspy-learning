# ChainOfThought, Field Descriptions & Typed Outputs

```python
import numpy
import dspy
from dotenv import load_dotenv
load_dotenv()

lm = dspy.LM('ollama_chat/llama3.1:8b', api_base='http://localhost:11434')
dspy.configure(lm=lm)
```

```python
from typing import Literal

class YesNo(dspy.Signature):
    """Answer the question. If the answer cannot be determined with confidence, say Uncertain."""

    question: str = dspy.InputField(desc="A yes/no question")
    answer: Literal["True", "False", "Uncertain"] = dspy.OutputField(desc="True, False, or Uncertain if the answer cannot be determined")

cot = dspy.ChainOfThought(YesNo)
```

```python
result = cot(question="The smallest country in the second largest continent is North America.")

print("Reasoning:", result.reasoning)
print("Answer:", result.answer)
print("Type:", type(result.answer))
```
