# Signatures & Modules

A **Signature** declares what goes in and what comes out of an LLM call.  
A **Module** defines the strategy for making that call. `dspy.Predict` is the simplest — one LLM call, no extra reasoning.

```python
class BasicQA(dspy.Signature):
    """Answer the question in a single, concise sentence. Your answer must be longer than a word."""
    question: str = dspy.InputField()
    answer: str = dspy.OutputField()

predict = dspy.Predict(BasicQA)
result = predict(question="What is the capital of France?")
print(result.answer)
```
